```python
import pandas as pd

df = pd.read_csv("/content/Dataset .csv")
print(df.head())
```

```python
df['Cuisines']=df['Cuisines'].fillna('Unknown')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
tfidf_matrix=tfidf.fit_transform(df['Cuisines'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(tfidf_matrix)
```

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

def recommend_restaurants(cuisine,price_range,city,top_n=5):
  user_input=cuisine+" "+str(price_range)+" "+city
  user_vector=tfidf.transform([user_input])
  similarity_scores=cosine_similarity(user_vector,tfidf_matrix).flatten()
  # Ensure similarity_scores is at least a 1D array to prevent IndexError with scalar
  similarity_scores = np.atleast_1d(similarity_scores)
  # Correctly get the top_n indices (last top_n elements) and reverse them
  top_indices=similarity_scores.argsort()[-top_n:][::-1]
  return df.iloc[top_indices][["Restaurant Name","Cuisines",'Price range',"City"]]
```

```python
recommend_restaurants(
    cuisine="Japanese",
    price_range=3,
    city="Makati City"
)
```

```python
recommend_restaurants(
    "Chinese",
    2,
    "Bangalore"
)
```

```python
recommend_restaurants(
    "Indian",
    4,
    "New Delhi"
)
```
